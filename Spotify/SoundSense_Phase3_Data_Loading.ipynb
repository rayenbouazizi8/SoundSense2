Phase 3 – Clustering

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from sklearn.manifold import TSNE
from sklearn.decomposition import PCA
import plotly.express as px
```

K-Means Clustering on genre data

```python
genre_features = [
    'acousticness',
    'danceability',
    'energy',
    'instrumentalness',
    'liveness',
    'valence'
]
```

```python
#Scale
scaler = StandardScaler()
genre_scaled = scaler.fit_transform(genre_data[genre_features])
```

Optimal K for Genre Clustering

```python
range_n_clusters = range(2, 21)
silhouette_scores = []

for k in range_n_clusters:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(genre_scaled)
    score = silhouette_score(genre_scaled, labels)
    silhouette_scores.append(score)

plt.figure(figsize=(8,5))
plt.plot(range_n_clusters, silhouette_scores, marker='o')
plt.title("Silhouette Score for Genre Clustering")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.show()
```

```python
#With_K=12(Silhouette score result)
kmeans_genre = KMeans(n_clusters=12, random_state=42)
genre_data['cluster'] = kmeans_genre.fit_predict(genre_scaled)
```

Visualize Genre clustering

```python
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
genre_tsne = tsne.fit_transform(genre_scaled)

genre_tsne_df = pd.DataFrame({
    'x': genre_tsne[:, 0],
    'y': genre_tsne[:, 1],
    'cluster': genre_data['cluster'],
    'genre': genre_data['genres']
})
px.scatter(
    genre_tsne_df,
    x='x',
    y='y',
    color='cluster',
     hover_data=['genre'],
    title='t-SNE Visualization of Genre Clusters'
)
```

K-Means Clustering on Song Data

```python
song_features = [
    'acousticness',
    'danceability',
    'energy',
    'instrumentalness',
    'liveness',
    'valence',
    'loudness',
    'tempo'
]
```

```python
#scale
song_scaled =scaler.fit_transform(data[song_features])
```

Silhouette for Song Clustering

```python
import numpy as np
sample_size = 1000
random_indices = np.random.choice(song_scaled.shape[0], size=sample_size, replace=False)
song_sample = song_scaled[random_indices]


range_n_clusters = range(20, 30)
silhouette_scores_song = []

for k in range_n_clusters:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(song_sample)   # ✅ use sample
    score = silhouette_score(song_sample, labels)  # ✅ use sample
    silhouette_scores_song.append(score)

plt.figure(figsize=(8,5))
plt.plot(range_n_clusters, silhouette_scores_song, marker='o')
plt.title("Silhouette Score for Song Clustering")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.show()
```

```python
#Optimal k=25 From Silhouette
kmeans_song = KMeans(n_clusters=25, random_state=42)
data['cluster'] = kmeans_song.fit_predict(song_scaled)
```

Visualize Song Clusters using PCA

```python
pca = PCA(n_components=2, random_state=42)
song_pca = pca.fit_transform(song_scaled)

song_pca_df = pd.DataFrame({
    'x': song_pca[:, 0],
    'y': song_pca[:, 1],
    'cluster': data['cluster'],
    'song': data['name'],
    'artist': data['artists']
})
px.scatter(
    song_pca_df,
    x='x',
    y='y',
    color='cluster',
    hover_data=['song', 'artist'],
    title='PCA Visualization of Song Clusters'
)
```

Hierarchical Clustering of Songs

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering

# 1️⃣ Load data
data = pd.read_csv("data.csv")

# 2️⃣ Create sample
sample_size = 3000
random_indices = np.random.choice(len(data), sample_size, replace=False)
data_sample = data.iloc[random_indices].copy()

# 3️⃣ Select features
song_features = [
    'acousticness',
    'danceability',
    'energy',
    'instrumentalness',
    'liveness',
    'valence',
    'loudness',
    'tempo'
]

# 4️⃣ Scale sample
scaler = StandardScaler()
song_scaled_sample = scaler.fit_transform(data_sample[song_features])

# 5️⃣ Fit hierarchical clustering ON SAMPLE
agg = AgglomerativeClustering(n_clusters=15)
data_sample['agg_cluster'] = agg.fit_predict(song_scaled_sample)

# 6️⃣ Check result
data_sample['agg_cluster'].value_counts()
```

Visualize with PCA

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2, random_state=42)
song_pca = pca.fit_transform(song_scaled_sample)


pca_df = pd.DataFrame({
    'x': song_pca[:, 0],
    'y': song_pca[:, 1],
    'cluster': data_sample['agg_cluster'],
    'song': data_sample['name'],
    'artist': data_sample['artists']
})
import plotly.express as px

fig = px.scatter(
    pca_df,
    x='x',
    y='y',
    color='cluster',
    hover_data=['song', 'artist'],
    title='Hierarchical Clustering (Agglomerative) - PCA Visualization'
)

fig.show()
```

Spectral Clustering

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Load data
data = pd.read_csv("data.csv")

# Create sample of 3000 songs
sample_size = 3000
random_indices = np.random.choice(len(data), sample_size, replace=False)
data_sample = data.iloc[random_indices].copy()

# Select features
song_features = [
    'acousticness',
    'danceability',
    'energy',
    'instrumentalness',
    'liveness',
    'valence',
    'loudness',
    'tempo'
]

# Scale sample
scaler = StandardScaler()
song_scaled_sample = scaler.fit_transform(data_sample[song_features])
```

Visualisa

```python
from sklearn.cluster import SpectralClustering

spectral = SpectralClustering(
    n_clusters=15,
    affinity='nearest_neighbors',
    random_state=42
)

data_sample['spectral_cluster'] = spectral.fit_predict(song_scaled_sample)
```

```python
data_sample.columns
```

```python
from sklearn.decomposition import PCA
import plotly.express as px

pca = PCA(n_components=2, random_state=42)
song_pca = pca.fit_transform(song_scaled_sample)

pca_spectral_df = pd.DataFrame({
    'x': song_pca[:, 0],
    'y': song_pca[:, 1],
    'cluster': data_sample['spectral_cluster'],
    'song': data_sample['name'],
    'artist': data_sample['artists']
})

fig = px.scatter(
    pca_spectral_df,
    x='x',
    y='y',
    color='cluster',
    hover_data=['song', 'artist'],
    title='Spectral Clustering (3000 Sample)'
)

fig.update_traces(marker=dict(size=5))
fig.show()
```

In this phase, we applied K-Means, Hierarchical (Agglomerative), and Spectral Clustering on a sample of 3000 songs using key Spotify audio features.
The visualizations show that all methods were able to segment songs into meaningful groups based on musical characteristics. However, Spectral Clustering produced better-separated and more structured clusters, especially in complex and overlapping regions of the feature space. This suggests that musical data may not always follow simple spherical patterns (as assumed by K-Means), and Spectral Clustering captures nonlinear relationships more effectively.
For genre clustering, we observed that genres with similar sound characteristics were grouped together, confirming that Spotify audio features reflect real musical similarities. This validates the consistency between audio-based clustering and genre-based categorization.
