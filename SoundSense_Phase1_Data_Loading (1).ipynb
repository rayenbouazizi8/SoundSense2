Phase 3 – Clustering

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from sklearn.manifold import TSNE
from sklearn.decomposition import PCA
import plotly.express as px
```

K-Means Clustering on genre data

```python
genre_features = [
    'acousticness',
    'danceability',
    'energy',
    'instrumentalness',
    'liveness',
    'valence'
]
```

```python
#Scale
scaler = StandardScaler()
genre_scaled = scaler.fit_transform(genre_data[genre_features])
```

Optimal K for Genre Clustering

```python
range_n_clusters = range(2, 21)
silhouette_scores = []

for k in range_n_clusters:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(genre_scaled)
    score = silhouette_score(genre_scaled, labels)
    silhouette_scores.append(score)

plt.figure(figsize=(8,5))
plt.plot(range_n_clusters, silhouette_scores, marker='o')
plt.title("Silhouette Score for Genre Clustering")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.show()
```

```python
#With_K=12(Silhouette score result)
kmeans_genre = KMeans(n_clusters=12, random_state=42)
genre_data['cluster'] = kmeans_genre.fit_predict(genre_scaled)
```

Visualize Genre clustering

```python
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
genre_tsne = tsne.fit_transform(genre_scaled)

genre_tsne_df = pd.DataFrame({
    'x': genre_tsne[:, 0],
    'y': genre_tsne[:, 1],
    'cluster': genre_data['cluster'],
    'genre': genre_data['genres']
})
px.scatter(
    genre_tsne_df,
    x='x',
    y='y',
    color='cluster',
     hover_data=['genre'],
    title='t-SNE Visualization of Genre Clusters'
)
```

Silhouette for Song Clustering

```python
import numpy as np
sample_size = 1000
random_indices = np.random.choice(song_scaled.shape[0], size=sample_size, replace=False)
song_sample = song_scaled[random_indices]


range_n_clusters = range(20, 30)
silhouette_scores_song = []

for k in range_n_clusters:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(song_sample)   # ✅ use sample
    score = silhouette_score(song_sample, labels)  # ✅ use sample
    silhouette_scores_song.append(score)

plt.figure(figsize=(8,5))
plt.plot(range_n_clusters, silhouette_scores_song, marker='o')
plt.title("Silhouette Score for Song Clustering")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.show()
```

```python
# Optimal k=25 From Silhouette
# The error "ValueError: Input X contains NaN" indicates that the 'song_scaled' array,
# which is used for KMeans, contains missing (NaN) values.
# This is likely due to missing values in the original 'data' DataFrame for the selected 'song_features'.

# To fix this, we first need to handle the missing values in 'data' before scaling.
# We will drop rows that contain NaN values in any of the 'song_features'.

# Drop rows with NaN values in the song features directly from the 'data' DataFrame
data.dropna(subset=song_features, inplace=True)

# Re-initialize the scaler and re-scale the cleaned data
scaler = StandardScaler()
song_scaled = scaler.fit_transform(data[song_features])

# Now perform KMeans clustering on the cleaned and scaled data
kmeans_song = KMeans(n_clusters=25, random_state=42, n_init='auto')
data['cluster'] = kmeans_song.fit_predict(song_scaled)
```

```python
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
from collections import defaultdict
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import cosine_similarity
```

## Spotify API Credentials
We use Spotify Developer credentials (Client ID + Client Secret) to fetch audio features when a song is not available in the local dataset.

```python
# Replace with your Spotify Developer credentials
CLIENT_ID = "YOUR_CLIENT_ID"
CLIENT_SECRET = "YOUR_CLIENT_SECRET"

sp = spotipy.Spotify(
    auth_manager=SpotifyClientCredentials(
        client_id=CLIENT_ID,
        client_secret=CLIENT_SECRET
    )
)
```

```python
song_features = [
    'acousticness',
    'danceability',
    'energy',
    'instrumentalness',
    'liveness',
    'valence',
    'loudness',
    'tempo'
]
```

```python
def find_song(name, year, spotify_data):
    song_row = spotify_data[(spotify_data['name'] == name) & (spotify_data['year'] == year)]
    if len(song_row) == 0:
        return None
    return song_row.iloc[0]
```

Get song data

```python
def get_song_data(song, spotify_data):
    # 1) Try dataset
    song_data = find_song(song['name'], song['year'], spotify_data)
    if song_data is not None:
        return song_data

    # 2) Fallback: Spotify API
    results = sp.search(q=f"track:{song['name']} year:{song['year']}", type="track", limit=1)

    if not results['tracks']['items']:
        return None

    track = results['tracks']['items'][0]
    audio_features = sp.audio_features(track['id'])[0]

    if audio_features is None:
        return None

    song_dict = {
        'name': song['name'],
        'year': song['year'],
        'artists': track['artists'][0]['name'],
        'acousticness': audio_features.get('acousticness'),
        'danceability': audio_features.get('danceability'),
        'energy': audio_features.get('energy'),
        'instrumentalness': audio_features.get('instrumentalness'),
        'liveness': audio_features.get('liveness'),
        'valence': audio_features.get('valence'),
        'loudness': audio_features.get('loudness'),
        'tempo': audio_features.get('tempo')
    }

    return pd.Series(song_dict)
```

Get Mean_Vector

```python
def get_mean_vector(song_list, spotify_data):
    vectors = []

    for song in song_list:
        song_data = get_song_data(song, spotify_data)
        if song_data is not None:
            vectors.append(song_data[song_features].values.astype(float))

    if len(vectors) == 0:
        return None

    return np.mean(vectors, axis=0)
```

Flatten list

```python
def flatten_dict_list(dict_list):
    flattened = defaultdict(list)
    for d in dict_list:
        for k, v in d.items():
            flattened[k].append(v)
    return flattened
```

Main recommend_song

```python
def recommend_songs(song_list, spotify_data, n_songs=10):
    # Get mean vector for user input
    song_center = get_mean_vector(song_list, spotify_data)
    if song_center is None:
        return "No valid songs found in dataset or Spotify API."

    # Scale dataset
    scaler = StandardScaler()
    scaled_data = scaler.fit_transform(spotify_data[song_features].fillna(0))

    # Scale center
    scaled_center = scaler.transform(song_center.reshape(1, -1))

    # Compute cosine similarity
    similarity = cosine_similarity(scaled_center, scaled_data)[0]

    # Get top indices (excluding identical matches if present)
    top_indices = np.argsort(similarity)[::-1][:n_songs]

    return spotify_data.iloc[top_indices][['name', 'year', 'artists']]
```
